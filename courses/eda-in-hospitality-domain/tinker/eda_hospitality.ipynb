# Hospitality Domain EDA: Revenue & Occupancy Analysis

This notebook performs exploratory data analysis across luxury hotel chains (AtliQ Grand) to evaluate **Occupancy %**, **Average Daily Rate (ADR)**, and **Revenue Per Available Room (RevPAR)** across Indian metropolitan markets.

```python
import pandas as pd
import numpy as np

# Synthesize representative hotel bookings data
cities = ['Mumbai', 'Delhi', 'Bangalore', 'Hyderabad']
room_types = ['Standard', 'Elite', 'Premium', 'Presidential']
platforms = ['Direct Online', 'Direct Offline', 'MakeMyTrip', 'Booking.com', 'Agoda']

print("Libraries loaded. Initializing Hospitality KPIs computation engine.")
```

## 1. Key Performance Indicator (KPI) Formulation

$$\text{RevPAR} = \text{ADR} \times \text{Occupancy Rate}$$

$$\text{Occupancy Rate (\%)} = \frac{\text{Successful Check-ins}}{\text{Total Capacity}} \times 100$$

```python
# Summary metrics by City
metrics_df = pd.DataFrame({
    'City': ['Mumbai', 'Delhi', 'Bangalore', 'Hyderabad'],
    'Total_Capacity': [12000, 9500, 14000, 8200],
    'Occupancy_Pct': [68.2, 74.5, 58.1, 62.4],
    'ADR_INR': [10557, 10872, 10155, 10336],
    'RevPAR_INR': [7200, 8100, 5900, 6450]
})

metrics_df.sort_values(by='RevPAR_INR', ascending=False)
```
